## 1. Model and question

The coupled state is bubble radius and speed, a fixed NASA9 gas mixture, material water temperatures and dynamic Au storage. The material coordinate is ξ=(r³−R³)/3. A single signed gas–water flow is used once in each subsystem with opposite signs. The paired comparison changes only the optical pulse within the same moving-water model. Spherical averaging does not resolve the finite-particle local hot spot.

## 6. Formal extremes and interpretation

The 1.15 bar trajectory completes five numerical cycles but remains a formal fixed-composition scenario: prescribed far-field absolute pressure becomes negative during part of a cycle, gas pressure reaches about 1.30×10⁸ Pa, and fixed composition plus the mechanical and incompressible-transport closures have unresolved physical applicability. Negative liquid pressure alone is not a mathematical proof of impossibility; metastable tension, cavitation and its stability are separate physical questions not resolved here. The 1.00 bar result remains conditional on the reduced model and unverified retention at a 100 nm gap.

In the matched static comparisons, peak Au temperature rises in MR7 are only approximately 21–30% of the resolved finite-Au MR6 values. Agreement in integrated gas-energy delivery (differences about 0.04–1.68%) does not validate the local Au temperature field. Consequently, melting, boiling and phase-transition thresholds in MR7 are diagnostic references and cannot establish physical admissibility. The static discrepancy is not a universal error bound for the moving model.

Dry fixed-composition 1.3 bar baselines remain formal at 6263.579 K (air) and 22808.716 K (ground-state constant-Cv Ar). Their pulse protocol differs from the nominal wet case. They do not supply the missing wet trajectory. The wet 6000 K stop is a coefficient boundary, not a physical temperature ceiling; the five-cycle global difference is undefined and no unknown tail is filled. No experimental temperature gain or fusion yield is established.
