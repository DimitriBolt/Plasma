## 1. Model and question

The coupled state is bubble radius and speed, a fixed NASA9 gas mixture, material water temperatures and dynamic Au storage. The material coordinate is ξ=(r³−R³)/3. A single signed gas–water flow is used once in each subsystem with opposite signs. The paired comparison changes only the optical pulse within the same moving-water model. Spherical averaging does not resolve the finite-particle local hot spot.

## 2. Primary paired result

One 10 ns, 532 nm pulse, 10 J/m², one 20 nm-radius Au particle, prescribed 100 nm gap; five acoustic periods where complete. The pulse center is five microseconds before the selected no-laser radius event.

These are cancellation-controlled nonlinear increments. Absolute baseline accuracy remains lower than the displayed increment resolution. The independent full Radau pair at 1.0 bar gives 1.4073 µK. The wet 1.3 bar case stops at the 6000 K coefficient boundary; no unknown tail is extrapolated.

## 5. Verification and reduction error

At 320 cells, matched stationary integrated gas delivery differs from MR6 by approximately 0.04–1.68%, while radial Au temperature rises are only 21–30% of the finite-Au reference. These discrepancies are different observables; no fitted correction is applied. The exact Robin point-source control, clamped MR4 regression, two stiff methods, finite increments, tangent responses, domain extension, mesh/time refinement and independent energy quadrature are retained. ## 6. Formal extremes and interpretation

Dry fixed-composition 1.3 bar baselines reach about 6264 K (air) and 22809 K (ground-state constant-Cv Ar). Their pulse protocol differs from the nominal wet case. They are formal sensitivities, with missing phase/chemical channels, nonideal EOS, shocks and local particle physics. The 6000 K wet boundary is a coefficient limit, not a physical ceiling. No experimental temperature gain or fusion yield is established.
